#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# Select the variables relevant to our research question
march_madness_tidy <- womens_march_madness |>
    select(reg_wins_pct, seed, tourney_wins) |>

    # Create a categorical variable for whether a team won at least one tournament game
    mutate(won_tourney_game = as_factor(tourney_wins > 0)) |>

    # Remove the original tourney_wins column
    select(-tourney_wins) |>

    # Remove rows with tournament seed of 0
    filter(seed != 0) |>

    # Remove rows with missing values
    drop_na()

# Display the first 6 rows
head(march_madness_tidy)
```

```r
march_madness_bar <- march_madness_tidy |>
    ggplot(aes(x = won_tourney_game)) +
    geom_bar(aes(y = after_stat(count / sum(count))),
             fill = "steelblue") +
    labs(
        title = "Proportion of Teams by Tournament Outcome",
        x = "Won at Least One Tournament Game",
        y = "Proportion of Teams"
    ) +
    scale_y_continuous(labels = scales::percent)

march_madness_bar
```

Visualization 1: Regular Season Win Percentage by Tournament Outcome

This boxplot compares regular season win percentages between teams that won at least one tournament game and those that did not. By comparing the medians and distributions of both groups, the visualization helps investigate whether regular season performance is associated with tournament success. This is relevant to the research question because it helps evaluate regular season win percentage as a potential predictor in the classification model.

```r
march_madness_boxplot <- march_madness_tidy |>
    ggplot(aes(x = won_tourney_game, y = reg_wins_pct)) +
    geom_boxplot(fill = "steelblue") +
    labs(
        title = "Regular Season Win Percentage by Tournament Outcome",
        x = "Won at Least One Tournament Game",
        y = "Regular Season Win Percentage (%)")

# Display the boxplot
march_madness_boxplot
```

Visualization 2: Tournament Win Rate by Seed

This line graph shows the proportion of teams that won at least one tournament game for each tournament seed. It allows for an examination of how tournament success rates change across different seed rankings and whether lower-numbered seeds tend to perform better. This is relevant to the research question because it helps determine whether tournament seed is a useful predictor of tournament success.

```r
march_madness_seed_plot <- march_madness_tidy |>
    ggplot(aes(x = as_factor(seed), fill = won_tourney_game)) +
    geom_bar(position = "fill") +
    labs(
        title = "Tournament Outcomes by Tournament Seed",
        x = "Tournament Seed",
        y = "Proportion of Teams",
        fill = "Won at Least One Game"
    ) +
    scale_y_continuous(labels = scales::percent)

march_madness_seed_plot
```
